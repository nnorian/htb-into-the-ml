```python
from fastmcp import Client

TARGET = "http://154.57.164.82:35239/mcp/"
client = Client(TARGET)

async def main():
    async with client:
        resources = await client.list_resources()
        templates = await client.list_resource_templates()
        tools = await client.list_tools()

        print("Resources:")
        for r in resources:
            print("***", r.name, "->", (r.description or "").strip())

        print("-"*50, "\nResource Templates:")
        for t in templates:
            print("***", t.uriTemplate, "->", (t.description or "").strip())

        print("-"*50, "\nTools:")
        for t in tools:
            params = list(t.inputSchema.get("properties", {}).keys())
            print(f"*** {t.name}({','.join(params)}) -> {(t.description or '').strip()}")

await main()
```

```python
async def read(uri):
    async with client:
        try:
            obj = await client.read_resource(uri)
            print(obj[0].text)
        except Exception as e:
            print(f"[-] {e}")

async def call(name, args):
    async with client:
        try:
            obj = await client.call_tool(name, args)
            print(obj.content[0].text)
        except Exception as e:
            print(f"[-] {e}")
```
